# Retail Sales EDA Project
**Tools:** Python, Pandas, NumPy, Matplotlib, Seaborn  
**Goal:** Explore retail sales data, clean data quality issues, visualize trends, and summarize business insights.

> This project includes a sample CSV file in the same folder, so it can run without downloading a dataset.

## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
```

## 2. Load the dataset

```python
df = pd.read_csv("retail_sales_sample.csv", parse_dates=["Order_Date"])
print("Dataset shape:", df.shape)
display(df.head())
```

## 3. Understand the data

```python
print("Rows and columns:", df.shape)
print("\\nColumn names:", df.columns.tolist())
print("\\nData types:")
display(df.dtypes.to_frame("dtype"))
print("\\nSummary statistics:")
display(df.describe(include="all").T)
```

## 4. Data quality checks

```python
print("Missing values per column:")
display(df.isna().sum().to_frame("missing_count"))
print("Duplicate rows:", df.duplicated().sum())
```

## 5. Clean the data

```python
df_clean = df.copy()
df_clean = df_clean.drop_duplicates()
df_clean["Region"] = df_clean["Region"].fillna("Unknown")
df_clean["Profit"] = df_clean["Profit"].fillna(df_clean["Profit"].median())
df_clean["Order_Date"] = pd.to_datetime(df_clean["Order_Date"], errors="coerce")
df_clean = df_clean.dropna(subset=["Order_Date", "Sales", "Quantity"])
print("Before cleaning:", df.shape)
print("After cleaning:", df_clean.shape)
print("Remaining missing values:", int(df_clean.isna().sum().sum()))
```

## 6. Key performance indicators (KPIs)

```python
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order_ID"].nunique()
avg_order_sales = df_clean.groupby("Order_ID")["Sales"].sum().mean()

print(f"Total sales: ₹{total_sales:,.2f}")
print(f"Total profit: ₹{total_profit:,.2f}")
print(f"Unique orders: {total_orders}")
print(f"Average sales per order: ₹{avg_order_sales:,.2f}")
```

## 7. Sales by category

```python
sales_by_category = df_clean.groupby("Category")["Sales"].sum().sort_values(ascending=False)
display(sales_by_category.to_frame("Total_Sales"))

sales_by_category.plot(kind="bar", figsize=(8, 4), title="Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## 8. Profit by region

```python
profit_by_region = df_clean.groupby("Region")["Profit"].sum().sort_values(ascending=False)
display(profit_by_region.to_frame("Total_Profit"))

profit_by_region.plot(kind="bar", figsize=(8, 4), title="Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 9. Monthly sales trend

```python
df_clean["Month"] = df_clean["Order_Date"].dt.to_period("M").astype(str)
monthly_sales = df_clean.groupby("Month")["Sales"].sum()

monthly_sales.plot(figsize=(12, 4), marker="o", title="Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

## 10. Sales vs profit

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x="Sales", y="Profit", hue="Category")
plt.title("Sales vs Profit by Category")
plt.tight_layout()
plt.show()
```

## 11. Correlation heatmap

```python
numeric_cols = df_clean.select_dtypes(include=np.number)
plt.figure(figsize=(7, 4))
sns.heatmap(numeric_cols.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Between Numeric Columns")
plt.tight_layout()
plt.show()
```

## 12. Business questions

```python
print("Highest sales category:", sales_by_category.idxmax())
print("Most profitable region:", profit_by_region.idxmax())
print("Lowest sales category:", sales_by_category.idxmin())
print("Total sales:", round(total_sales, 2))
print("Total profit:", round(total_profit, 2))
```
